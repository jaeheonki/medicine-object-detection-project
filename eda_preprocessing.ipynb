## 1. 환경 설정 및 데이터 로드

````python
import os
import json
import random
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.patches as patches
from pathlib import Path
from PIL import Image, ImageFilter, ImageEnhance
from collections import Counter
import cv2
import albumentations as A
from albumentations.pytorch import ToTensorV2

# 경로 설정 — 환경에 맞게 수정
DATA_DIR = Path('./sprint_ai_project1_data')
TRAIN_IMG_DIR  = DATA_DIR / 'train_images'
TRAIN_ANN_DIR  = DATA_DIR / 'train_annotations'
TEST_IMG_DIR   = DATA_DIR / 'test_images'

plt.rcParams['font.family'] = 'AppleGothic'
plt.rcParams['axes.unicode_minus'] = False
plt.rcParams['figure.dpi']  = 120
print('✅ 환경 설정 완료')
````

````python
def load_all_annotations(ann_dir: Path) -> dict:
    """
    여러 JSON 파일로 분산된 COCO 포맷 어노테이션을
    하나의 딕셔너리로 병합합니다.
    """
    all_images      = []
    all_annotations = []
    all_categories  = {}

    for json_file in sorted(ann_dir.rglob('*.json')):
        with open(json_file, 'r', encoding='utf-8') as f:
            data = json.load(f)

        all_images.extend(data.get('images', []))
        all_annotations.extend(data.get('annotations', []))

        for cat in data.get('categories', []):
            all_categories[cat['id']] = cat

    return {
        'images':      all_images,
        'annotations': all_annotations,
        'categories':  list(all_categories.values())
    }

coco = load_all_annotations(TRAIN_ANN_DIR)

print(f'이미지 수      : {len(coco["images"]):,}')
print(f'어노테이션 수  : {len(coco["annotations"]):,}')
print(f'카테고리(클래스) 수: {len(coco["categories"]):,}')
````

````python
# ── 2-4. 샘플 이미지 + 바운딩 박스 시각화 ────────────────────────
def visualize_sample(image_id: int, n_samples: int = 4):
    img_info_list = [img for img in coco['images'] if img['id'] == image_id]
    if not img_info_list:
        print(f'image_id {image_id} 없음')
        return

    img_info = img_info_list[0]
    img_path = TRAIN_IMG_DIR / img_info['file_name']
    img      = Image.open(img_path).convert('RGB')

    anns = [a for a in coco['annotations'] if a['image_id'] == image_id]

    fig, ax = plt.subplots(1, 1, figsize=(7, 7))
    ax.imshow(img)

    colors = ['tomato', 'dodgerblue', 'limegreen', 'orange']
    for i, ann in enumerate(anns):
        x, y, w, h = ann['bbox']
        rect = patches.Rectangle((x, y), w, h,
                                  linewidth=2, edgecolor=colors[i % len(colors)], facecolor='none')
        ax.add_patch(rect)
        label = cat_id_to_name.get(ann['category_id'], str(ann['category_id']))
        ax.text(x, y - 4, label, fontsize=8, color=colors[i % len(colors)],
                bbox=dict(facecolor='white', alpha=0.6, pad=1, edgecolor='none'))

    ax.set_title(f"image_id={image_id} | 알약 {len(anns)}개 | {img_info.get('dl_name', '')}")
    ax.axis('off')
    plt.tight_layout()
    plt.show()

# 랜덤 샘플 4장 시각화
sample_ids = random.sample([img['id'] for img in coco['images']], 4)
for sid in sample_ids:
    visualize_sample(sid)
````

## 4. 전처리 전략 구현

### 전략 요약

```
[환경 극복]        배경 합성 / Color Jittering / 가우시안 노이즈·블러
[반쪽 알약 대응]   Random Erasing / Cutout
[정교한 인식]      Sharpening / 고해상도 리사이즈
```

````python
# ── 4-1. Albumentations 증강 파이프라인 ───────────────────────────
#
# 모델링 팀은 이 transform을 DataLoader에 그대로 연결하면 됩니다.
# 바운딩 박스 포맷: 'coco' = [x_min, y_min, width, height]

train_transform = A.Compose([

    # ① 고해상도 리사이즈 — mAP@0.75+ 기준 맞추기 위해 디테일 보존
    A.LongestMaxSize(max_size=1024),
    A.PadIfNeeded(min_height=1024, min_width=1024,
                  border_mode=cv2.BORDER_CONSTANT, value=0),

    # ② 환경 편향 극복 — 조명 / 색상 변조
    A.ColorJitter(
        brightness=0.3,   # 밝기 ±30%
        contrast=0.3,     # 대비 ±30%
        saturation=0.2,   # 채도 ±20%
        hue=0.05,         # 색조 미세 변화
        p=0.7
    ),

    # ③ 환경 편향 극복 — 노이즈 & 블러 (폰카 흔들림 시뮬레이션)
    A.OneOf([
        A.GaussNoise(std_range=(0.01, 0.05)),
        A.MotionBlur(blur_limit=5),
        A.GaussianBlur(blur_limit=3),
    ], p=0.4),

    # ④ 반쪽 알약 대응 — 일부 영역 강제 삭제
    A.CoarseDropout(
        max_holes=4,
        max_height=80,
        max_width=80,
        min_holes=1,
        fill_value=0,
        p=0.4
    ),

    # ⑤ 기하학적 변환 — 회전 / 플립 (알약은 방향 무관)
    A.HorizontalFlip(p=0.5),
    A.VerticalFlip(p=0.3),
    A.Rotate(limit=30, p=0.5),

    # ⑥ 정교한 인식 — 샤프닝 (각인 문자 선명도 향상)
    A.Sharpen(alpha=(0.2, 0.5), lightness=(0.5, 1.0), p=0.3),

    # ⑦ 정규화
    A.Normalize(mean=[0.485, 0.456, 0.406],
                std=[0.229, 0.224, 0.225]),
    ToTensorV2(),

], bbox_params=A.BboxParams(
    format='coco',
    label_fields=['category_ids'],
    min_visibility=0.3       # bbox 30% 미만으로 잘리면 제거
))

# 검증/테스트용 (증강 없이 리사이즈 + 정규화만)
val_transform = A.Compose([
    A.LongestMaxSize(max_size=1024),
    A.PadIfNeeded(min_height=1024, min_width=1024,
                  border_mode=cv2.BORDER_CONSTANT, fill=0),
    A.Normalize(mean=[0.485, 0.456, 0.406],
                std=[0.229, 0.224, 0.225]),
    ToTensorV2(),
], bbox_params=A.BboxParams(format='coco', label_fields=['category_ids']))

print(' 증강 파이프라인 정의 완료')
print(f'   train_transform 단계 수: {len(train_transform.transforms)}')
````

````python
# ── 4-2. 배경 합성 유틸 ───────────────────────────────────────────
#
# 고정 회색 배경 대신 나무 책상, 유리 테이블 등 다양한 배경을 합성합니다.
# bg_dir 에 배경 이미지들을 넣어두면 랜덤 적용됩니다.

def swap_background(pill_img: np.ndarray,
                    bg_dir: Path,
                    bg_thresh: int = 200) -> np.ndarray:
    """
    pill_img : 원본 알약 이미지 (H, W, 3) — BGR
    bg_dir   : 대체할 배경 이미지 폴더
    bg_thresh: 배경으로 판단할 밝기 임계값 (연회색 배경 기준)
    """
    bg_paths = list(bg_dir.glob('*.jpg')) + list(bg_dir.glob('*.png'))
    if not bg_paths:
        return pill_img  # 배경 없으면 원본 반환

    h, w = pill_img.shape[:2]

    # 배경 마스크 생성 (밝고 채도 낮은 영역 = 회색 배경)
    gray    = cv2.cvtColor(pill_img, cv2.COLOR_BGR2GRAY)
    _, mask = cv2.threshold(gray, bg_thresh, 255, cv2.THRESH_BINARY)
    mask    = cv2.erode(mask, np.ones((3, 3)), iterations=1)

    # 랜덤 배경 불러와서 리사이즈
    bg = cv2.imread(str(random.choice(bg_paths)))
    bg = cv2.resize(bg, (w, h))

    # 합성
    mask_3ch = cv2.merge([mask, mask, mask])
    result   = np.where(mask_3ch == 255, bg, pill_img)
    return result

print('✅ swap_background 함수 정의 완료')
print('   사용법: result = swap_background(cv2_img, Path("./backgrounds"))')
````

## 5. 증강 기법 시각화

각 증강 기법이 실제로 이미지를 어떻게 바꾸는지 모델링 팀이 한눈에 확인할 수 있도록 시각화합니다.

````python
# ── 5-1. 증강 기법별 Before / After 비교 ─────────────────────────

def show_augmentation_grid(image_id: int):
    """하나의 원본 이미지에 각 증강 기법을 적용해 나란히 보여줍니다."""
    img_info = next(img for img in coco['images'] if img['id'] == image_id)
    img_path = TRAIN_IMG_DIR / img_info['file_name']
    img_np   = np.array(Image.open(img_path).convert('RGB'))

    augmentations = [
        ('원본',                    None),
        ('Color Jitter\n(조명 변화)', A.ColorJitter(brightness=0.4, contrast=0.4,
                                                   saturation=0.4, hue=0.1, p=1.0)),
        ('Gaussian Noise\n(노이즈)', A.GaussNoise(var_limit=(40, 80), p=1.0)),
        ('Motion Blur\n(손떨림)',    A.MotionBlur(blur_limit=9, p=1.0)),
        ('Coarse Dropout\n(반쪽 대응)', A.CoarseDropout(num_holes_range=(1, 8), hole_height_range=(1, 8),
                                                       hole_width_range=(1, 8), fill=0)),
        ('Sharpen\n(각인 선명화)',   A.Sharpen(alpha=(0.5, 0.8), lightness=(0.8, 1.0), p=1.0)),
        ('Rotate 30°\n(각도 변화)', A.Rotate(limit=30, p=1.0)),
        ('Horizontal Flip',         A.HorizontalFlip(p=1.0)),
    ]

    fig, axes = plt.subplots(2, 4, figsize=(16, 8))
    axes = axes.flatten()

    for ax, (title, aug) in zip(axes, augmentations):
        if aug is None:
            out = img_np
        else:
            out = A.Compose([aug])(image=img_np)['image']
        ax.imshow(out)
        ax.set_title(title, fontsize=10)
        ax.axis('off')

    plt.suptitle(f'증강 기법 비교 | image_id={image_id}', fontsize=13, y=1.01)
    plt.tight_layout()
    plt.show()

# 실행: 원하는 image_id 로 변경
show_augmentation_grid(image_id=sample_ids[0])
````

````python
# ── 5-2. 증강 후 바운딩 박스 유지 확인 ────────────────────────────
#
# 증강 후 bbox가 올바르게 따라오는지 검증하는 셀입니다.

def verify_bbox_after_aug(image_id: int):
    img_info = next(img for img in coco['images'] if img['id'] == image_id)
    img_path = TRAIN_IMG_DIR / img_info['file_name']
    img_np   = np.array(Image.open(img_path).convert('RGB'))

    anns     = [a for a in coco['annotations'] if a['image_id'] == image_id]
    bboxes   = [a['bbox'] for a in anns]
    cat_ids  = [a['category_id'] for a in anns]

    aug_for_bbox = A.Compose([
        A.Rotate(limit=30, p=1.0),
        A.HorizontalFlip(p=1.0),
        A.ColorJitter(brightness=0.3, contrast=0.3, p=1.0),
    ], bbox_params=A.BboxParams(format='coco', label_fields=['category_ids'],
                                min_visibility=0.3))

    result      = aug_for_bbox(image=img_np, bboxes=bboxes, category_ids=cat_ids)
    aug_img     = result['image']
    aug_bboxes  = result['bboxes']

    fig, axes = plt.subplots(1, 2, figsize=(12, 5))
    colors    = ['tomato', 'dodgerblue', 'limegreen', 'orange']

    for ax, (image, boxes, title) in zip(axes, [
        (img_np,  bboxes,     '원본 + bbox'),
        (aug_img, aug_bboxes, '증강 후 + bbox'),
    ]):
        ax.imshow(image)
        for i, (x, y, w, h) in enumerate(boxes):
            rect = patches.Rectangle((x, y), w, h, linewidth=2,
                                     edgecolor=colors[i % len(colors)], facecolor='none')
            ax.add_patch(rect)
        ax.set_title(title)
        ax.axis('off')

    plt.suptitle(f'bbox 보존 검증 | 원본 {len(bboxes)}개 → 증강 후 {len(aug_bboxes)}개', fontsize=12)
    plt.tight_layout()
    plt.show()

verify_bbox_after_aug(image_id=sample_ids[1])
````

## 6. 클래스 불균형 분석 및 대응

소수 클래스를 파악하고 오버샘플링 가중치를 계산합니다.  
모델링 팀은 `class_weights`를 loss function 또는 DataLoader sampler에 직접 연결할 수 있습니다.

````python
# ── 6-1. 클래스 불균형 시각화 및 소수 클래스 추출 ───────────────────

class_counts = ann_df['category_id'].value_counts()
total_ann    = len(ann_df)
n_classes    = len(class_counts)

# 하위 20% = 소수 클래스로 정의
threshold    = class_counts.quantile(0.2)
minority_ids = class_counts[class_counts <= threshold].index.tolist()

fig, axes = plt.subplots(1, 2, figsize=(14, 4))

bar_colors = ['tomato' if i in minority_ids else 'steelblue'
              for i in class_counts.index]
axes[0].bar(range(len(class_counts)), class_counts.values,
            color=bar_colors, width=0.8)
axes[0].axhline(threshold, color='tomato', linestyle='--',
                label=f'소수 클래스 기준 ({threshold:.0f}개 이하)')
axes[0].set_title('클래스별 어노테이션 수 (빨강 = 소수 클래스)')
axes[0].set_xlabel('클래스 순위')
axes[0].set_ylabel('어노테이션 수')
axes[0].legend()

minority_counts = class_counts[class_counts.index.isin(minority_ids)]
axes[1].barh(
    [cat_id_to_name.get(i, str(i)) for i in minority_counts.index],
    minority_counts.values,
    color='tomato'
)
axes[1].set_title(f'소수 클래스 상세 ({len(minority_ids)}개)')
axes[1].set_xlabel('어노테이션 수')

plt.tight_layout()
plt.show()

print(f'전체 클래스 수  : {n_classes}')
print(f'소수 클래스 수  : {len(minority_ids)} (하위 20%)')
print(f'소수 클래스 ID  : {minority_ids[:10]} ...')
````

````python
# ── 6-2. WeightedRandomSampler용 샘플 가중치 계산 ──────────────────
#
# 모델링 팀: DataLoader(dataset, batch_size=8, sampler=weighted_sampler)

import torch
from torch.utils.data import WeightedRandomSampler

# 이미지별 대표 category_id (알약 여러 개면 첫 번째 기준)
img_to_cat = (
    ann_df.sort_values('image_id')
          .groupby('image_id')['category_id']
          .first()
)

# 클래스별 등장 빈도 역수 → 가중치
cat_freq   = img_to_cat.value_counts()
cat_weight = 1.0 / cat_freq

image_ids_ordered = [img['id'] for img in coco['images']]
sample_weights    = torch.tensor(
    [float(cat_weight.get(img_to_cat.get(iid, -1), 1.0)) for iid in image_ids_ordered],
    dtype=torch.float
)

weighted_sampler = WeightedRandomSampler(
    weights     = sample_weights,
    num_samples = len(sample_weights),
    replacement = True
)

print('✅ WeightedRandomSampler 생성 완료')
print(f'   샘플 수       : {len(sample_weights):,}')
print(f'   최대 가중치   : {sample_weights.max():.4f}  (소수 클래스)')
print(f'   최소 가중치   : {sample_weights.min():.4f}  (다수 클래스)')
print()
print('📌 모델링 팀 사용법:')
print('   DataLoader(dataset, batch_size=8, sampler=weighted_sampler)')
````

## 7. 메타데이터 활용 가능성 분석

비즈니스 UI 연동을 위해 핵심 JSON 필드의 결측률을 확인합니다.  
`dl_material`(성분), `line_front`(분할선), `di_etc_otc_code`(전문/일반) 등이 얼마나 채워져 있는지 파악합니다.

````python
# ── 7-1. UI 연동 핵심 필드 채움률 분석 ─────────────────────────────

img_df = pd.DataFrame(coco['images'])

ui_fields = [
    ('dl_material',      '성분명'),
    ('dl_material_en',   '성분명(영어)'),
    ('di_etc_otc_code',  '전문/일반 구분'),
    ('line_front',       '앞면 분할선'),
    ('line_back',        '뒷면 분할선'),
    ('dl_name',          '제품명'),
    ('dl_company',       '제조사'),
    ('chart',            '제형'),
    ('drug_shape',       '알약 모양'),
    ('color_class1',     '색상'),
    ('print_front',      '앞면 각인'),
    ('print_back',       '뒷면 각인'),
]

rows = []
for col, label in ui_fields:
    if col in img_df.columns:
        total     = len(img_df)
        filled    = img_df[col].notna() & (img_df[col].astype(str).str.strip() != '')
        fill_rate = filled.sum() / total * 100
        rows.append({'필드': col, '설명': label,
                     '채움률(%)': round(fill_rate, 1),
                     'UI활용': '✅' if fill_rate >= 80 else ('⚠️' if fill_rate >= 40 else '❌')})
    else:
        rows.append({'필드': col, '설명': label, '채움률(%)': 0.0, 'UI활용': '❌ (컬럼없음)'})

meta_df = pd.DataFrame(rows).sort_values('채움률(%)', ascending=False)
print(meta_df.to_string(index=False))

fig, ax = plt.subplots(figsize=(10, 5))
colors  = ['steelblue' if r >= 80 else ('orange' if r >= 40 else 'tomato')
           for r in meta_df['채움률(%)']]
ax.barh(meta_df['설명'], meta_df['채움률(%)'], color=colors)
ax.axvline(80, color='steelblue', linestyle='--', alpha=0.6, label='80% (활용 권장)')
ax.axvline(40, color='orange',    linestyle='--', alpha=0.6, label='40% (주의)')
ax.set_xlabel('채움률 (%)')
ax.set_title('UI 연동 핵심 필드 채움률\n파랑 ≥80% 활용가능 / 주황 ≥40% 주의 / 빨강 활용불가')
ax.legend()
plt.tight_layout()
plt.show()
````

````python
# ── 7-2. 분할선 & 전문/일반의약품 분포 (복용 가이드 UI 핵심) ──────

fig, axes = plt.subplots(1, 3, figsize=(15, 4))

for ax, (col, title) in zip(axes, [
    ('line_front',      '앞면 분할선 여부'),
    ('di_etc_otc_code', '전문/일반의약품'),
    ('drug_shape',      '알약 모양'),
]):
    if col in img_df.columns:
        counts = img_df[col].value_counts().head(8)
        ax.pie(counts.values, labels=counts.index,
               autopct='%1.1f%%', startangle=90)
        ax.set_title(title)
    else:
        ax.text(0.5, 0.5, f'컬럼 없음:\n{col}',
                ha='center', va='center', transform=ax.transAxes)
        ax.set_title(title)

plt.suptitle('복용 가이드 UI 연동 핵심 분포', fontsize=12)
plt.tight_layout()
plt.show()

if 'line_front' in img_df.columns:
    has_line = (
        img_df['line_front'].notna() &
        (img_df['line_front'].astype(str).str.strip() != '')
    )
    print(f'분할선 있는 약 비율: {has_line.mean()*100:.1f}%')
    print('→ 이 비율만큼 "반으로 잘라 드셔도 됩니다" 가이드 제공 가능')
````
