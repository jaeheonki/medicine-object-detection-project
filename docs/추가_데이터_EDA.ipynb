```python
import json
# 각 json 파일의 클래스 목록 확인
all_classes = set()
for json_file in Path(MERGED_ANN_DIR).glob("*.json"):
    with open(json_file, "r", encoding="utf-8") as f:
        data = json.load(f)
    for cat in data.get("raw_categories", []):
        all_classes.add(cat["name"])

print(f"전체 고유 클래스 수: {len(all_classes)}")
for c in sorted(all_classes):
    print(c)
```

```python
import json
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.ticker as ticker
from pathlib import Path
from collections import defaultdict

# ─── 데이터 로드 ────────────────────────────────────────────────────
def load_class_counts(merged_ann_dir):
    cat_id_to_name = {}
    cat_counts = defaultdict(int)

    for json_file in Path(merged_ann_dir).glob("*.json"):
        with open(json_file, "r", encoding="utf-8") as f:
            item = json.load(f)

        for cat in item.get("raw_categories", []):
            cat_id_to_name[cat["id"]] = cat["name"]

        for ann in item.get("merged_annotations", []):
            if ann.get("bbox_xyxy") is not None:
                cat_id = ann["raw_annotation"]["category_id"]
                cat_counts[cat_id] += 1

    rows = [
        {"category_id": cid, "name": cat_id_to_name.get(cid, str(cid)), "count": cnt}
        for cid, cnt in cat_counts.items()
    ]
    df = pd.DataFrame(rows).sort_values("count", ascending=False).reset_index(drop=True)
    return df

df = load_class_counts(MERGED_ANN_DIR)
print(f"전체 클래스 수: {len(df)}")
print(f"전체 어노테이션 수: {df['count'].sum()}")
print(df.head(10).to_string(index=False))
```

```python
%matplotlib inline
import matplotlib.pyplot as plt
import matplotlib.ticker as ticker

# ─── 시각화 ─────────────────────────────────────────────────────────
fig, axes = plt.subplots(1, 2, figsize=(22, max(10, len(df) * 0.25)))

# ── 왼쪽: 전체 클래스 가로 막대 그래프 ──────────────────────────────
colors = ["tomato" if c <= df["count"].quantile(0.2) else "steelblue" for c in df["count"]]

axes[0].barh(range(len(df)), df["count"], color=colors, height=0.8)
axes[0].axvline(df["count"].mean(), color="orange", linestyle="--", linewidth=1.5,
                label=f"평균 {df['count'].mean():.1f}개")
axes[0].axvline(df["count"].quantile(0.2), color="tomato", linestyle=":", linewidth=1.5,
                label=f"하위 20% 기준 ({df['count'].quantile(0.2):.0f}개 이하, 소수 클래스)")
axes[0].set_title(f"클래스별 어노테이션 수\n전체 {len(df)}개 클래스  (빨강 = 소수 클래스)",
                  fontsize=14, fontweight="bold", pad=12)
axes[0].set_ylabel("클래스명", fontsize=11)
axes[0].set_xlabel("어노테이션 수", fontsize=11)
axes[0].set_yticks(range(len(df)))
axes[0].set_yticklabels(df["name"], fontsize=9)
axes[0].invert_yaxis()   # 많은 순으로 위에서 아래로
axes[0].legend(fontsize=9)
axes[0].xaxis.set_major_locator(ticker.MaxNLocator(integer=True))

# ── 오른쪽: 상위 20개 + 하위 20개 비교 ─────────────────────────────
top20 = df.head(20)
bot20 = df.tail(20)

y = range(20)
height = 0.35

axes[1].barh([i + height/2 for i in y], top20["count"].values, height=height,
             color="steelblue", label="상위 20개 클래스")
axes[1].barh([i - height/2 for i in y], bot20["count"].values, height=height,
             color="tomato",   label="하위 20개 클래스")

axes[1].set_title(
    f"상위 20 vs 하위 20 클래스\n최대/최소 불균형 비율: {df['count'].iloc[0] / df['count'].iloc[-1]:.1f}배",
    fontsize=14, fontweight="bold", pad=12)
axes[1].set_ylabel("클래스 (상위 vs 하위)", fontsize=11)
axes[1].set_xlabel("어노테이션 수", fontsize=11)
axes[1].set_yticks(range(20))
axes[1].set_yticklabels(
    [f"{a}  /  {b}" for a, b in zip(top20["name"], bot20["name"])],
    fontsize=9
)
axes[1].invert_yaxis()
axes[1].legend(fontsize=10)
axes[1].xaxis.set_major_locator(ticker.MaxNLocator(integer=True))

plt.tight_layout(pad=3.0)
plt.show()
```
